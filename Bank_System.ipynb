```python
# ========================================
#        Bank System 🏦
# ========================================

class BankAccount:
    def __init__(self, account_holder, account_number, balance):
        self.account_holder = account_holder
        self.account_number = account_number
        self.__balance = balance  # Private variable

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"✅ {amount} deposited successfully!")
        else:
            print("❌ Deposit amount must be greater than 0!")

    def withdraw(self, amount):
        if amount <= 0:
            print("❌ Withdrawal amount must be greater than 0!")
        elif amount <= self.__balance:
            self.__balance -= amount
            print(f"✅ {amount} withdrawn successfully!")
        else:
            print("❌ Insufficient balance!")

    def check_balance(self):
        print(f"💰 Current Balance: {self.__balance}")
        return self.__balance

    def show_details(self):
        print(
            f"Account Holder: {self.account_holder} | "
            f"Account No: {self.account_number} | "
            f"Balance: {self.__balance}"
        )


# ========================================
#        Savings Account Class
# ========================================

class SavingsAccount(BankAccount):
    def __init__(self, account_holder, account_number, balance, interest_rate):
        super().__init__(account_holder, account_number, balance)
        self.interest_rate = interest_rate

    def add_interest(self):
        interest = self.check_balance() * self.interest_rate / 100
        self.deposit(interest)
        print(f"📈 Interest of {self.interest_rate}% added!")

    def show_details(self):
        print(
            f"[Savings] Account Holder: {self.account_holder} | "
            f"Account No: {self.account_number} | "
            f"Balance: {self.check_balance()} | "
            f"Interest Rate: {self.interest_rate}%"
        )


# ========================================
#              Main Program
# ========================================

print("========================================")
print("         Bank System 🏦")
print("========================================")


# Bank Account Object
account1 = BankAccount("Ali", 101, 1000)

account1.show_details()

account1.deposit(500)
account1.withdraw(200)

account1.check_balance()

account1.withdraw(1500)


print()

# Savings Account Object
account2 = SavingsAccount("Saad", 102, 5000, 10)

account2.show_details()

account2.deposit(1000)

account2.add_interest()

account2.check_balance()
```
