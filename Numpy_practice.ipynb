52. Consider a random vector with shape (100,2) representing coordinates, find point by point distances (★★☆)


```python
import numpy as np

X = np.random.random((100, 2))

distances = np.sqrt(np.sum(np.diff(X, axis=0)**2, axis=1))

print(distances)
```


62. Considering two arrays with shape (1,3) and (3,1), how to compute their sum using an iterator? (★★☆)

```python
import numpy as np

A = np.arange(3).reshape(3, 1)
B = np.arange(3).reshape(1, 3)

# Pass None as the third operand to allocate the output array automatically
it = np.nditer([A, B, None])

# Iterate and compute the element-wise sum
for x, y, z in it:
    z[...] = x + y

# Retrieve the result from the operands list
result = it.operands[2]

print("Array A:\n", A)
print("\nArray B:\n", B)
print("\nSum via Iterator:\n", result)
```
